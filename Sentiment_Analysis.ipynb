### Step 1: Prepare a Small, Balanced Few-Shot Dataset

We'll sample a small number of tweets for each sentiment class to create our few-shot training and validation sets. This simulates a scenario where only limited labeled data is available.

````python
import pandas as pd
from sklearn.model_selection import train_test_split

# Map sentiment labels to integers for the entire cleaned dataset
label_to_id = {label: i for i, label in enumerate(candidate_labels)}
id_to_label = {i: label for i, label in enumerate(candidate_labels)}
df_cleaned['labels'] = df_cleaned['Sentiment'].map(label_to_id)

print(f"Full cleaned dataset size: {len(df_cleaned)}")

# Split the entire cleaned dataset into training and validation (80-20 split)
# We use stratify to ensure that the class distribution is maintained in both sets.
train_df, val_df = train_test_split(df_cleaned, test_size=0.2, random_state=42, stratify=df_cleaned['labels'])

print(f"Training set size: {len(train_df)}")
print(f"Validation set size: {len(val_df)}")
````

### Step 3: Tokenize the Datasets

We'll convert the text content of our training and validation data into numerical tokens that the model can process. We'll also set the format for PyTorch, as the `Trainer` expects it.

````python
def tokenize_function(examples):
    # Truncation handles sequences longer than the model's maximum input size
    # Padding ensures all sequences have the same length for batch processing
    return tokenizer(examples['Tweet content'], truncation=True, padding='max_length', max_length=128)

# Convert pandas DataFrames to Hugging Face Dataset objects
train_dataset = Dataset.from_pandas(train_df[['Tweet content', 'labels']])
val_dataset = Dataset.from_pandas(val_df[['Tweet content', 'labels']])

# Apply tokenization to both datasets
tokenized_train_dataset = train_dataset.map(tokenize_function, batched=True)
tokenized_val_dataset = val_dataset.map(tokenize_function, batched=True)

# Remove the original text column and set the format for PyTorch
tokenized_train_dataset = tokenized_train_dataset.remove_columns(["Tweet content"])
tokenized_val_dataset = tokenized_val_dataset.remove_columns(["Tweet content"])

tokenized_train_dataset.set_format("torch")
tokenized_val_dataset.set_format("torch")
````

### Step 2: Generate Embeddings and Create a Vector Store

````python
def retrieve_and_generate(tweet_text, top_k=3):
    # 1. Retrieve relevant documents
    query_embedding = embedder.encode(tweet_text, convert_to_tensor=True).cpu().numpy().reshape(1, -1)
    distances, indices = index.search(query_embedding, top_k)

    retrieved_tweets = [corpus_sentences[idx] for idx in indices[0]]

    # 2. Construct a prompt for the generator to classify sentiment
    context = "\n".join(retrieved_tweets)
    rag_prompt = f"""
Given the following related tweets as context:
{context}

Classify the sentiment of the following tweet. Choose only one of these labels: {', '.join(candidate_labels)}
Tweet: {tweet_text}
Sentiment:
"""

    # 3. Generate a response using the LLM
    generation_config = GenerationConfig(
        max_new_tokens=50, # Keep the generated response concise to aid parsing
        pad_token_id=50256,
        num_return_sequences=1
    )
    output = generator(rag_prompt, generation_config=generation_config)
    generated_text = output[0]['generated_text']

    # Extract only the answer part, if the model re-generates the prompt
    if rag_prompt in generated_text:
        extracted_response = generated_text.replace(rag_prompt, '').strip()
    else:
        extracted_response = generated_text

    # Attempt to extract a valid sentiment label from the response
    predicted_sentiment = 'Unknown' # Default if no label is found
    for label in candidate_labels:
        if label.lower() in extracted_response.lower(): # Case-insensitive check
            predicted_sentiment = label
            break

    return predicted_sentiment, retrieved_tweets

# --- Example usage (updated) --- (this part is commented out for cleaner execution)
# query_tweet = "I forgot how great it is to play Borderlands. MAN I love this franchise."
# predicted_sentiment, retrieved_docs = retrieve_and_generate(query_tweet)

# print(f"Tweet: {query_tweet}")
# print("\nRetrieved Tweets:")
# for i, doc in enumerate(retrieved_docs):
#     print(f"{i+1}. {doc}")
# print("\nPredicted Sentiment (RAG):")
# print(predicted_sentiment)
````

````python
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = logits.argmax(axis=1)

    precision, recall, f1, _ = precision_recall_fscore_support(labels, predictions, average='weighted')
    acc = accuracy_score(labels, predictions)

    return {
        "accuracy": acc,
        "f1": f1,
        "precision": precision,
        "recall": recall
    }
````
